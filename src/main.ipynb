Task 0 : Loading the dataset

```python
import numpy as np
 
data = np.genfromtxt(
    '../dataset/deliveries.csv',
    delimiter=',',
    dtype=str,
    skip_header=1
)
 
print("Shape:", data.shape)
print("First row:", data[0])
 
match_ids = data[:, 0].astype(int)
over = data[:, 1].astype(int)
batting_team = data[:, 2]
batter = data[:, 6]
bowler = data[:, 7]
batsman_runs = data[:, 9].astype(int)
```

Taks 1 : Total runs per match


```python
unique_match_ids = np.unique(match_ids)

total_runs_per_match = [(int(match_id), int(batsman_runs[match_ids == match_id].sum())) for match_id in unique_match_ids]

print(total_runs_per_match[:5])
```

Task 2 : Top 5 Batters


```python
# Sum runs for each batter
top_5_batters = [
    (str(player), int(batsman_runs[batter == player].sum()))
    for player in np.unique(batter)
]

# Sort by total runs in descending order and keep the top 5
top_5_batters = sorted(top_5_batters, key=lambda x: x[1], reverse=True)[:5]

print(top_5_batters)
```

Task 3 : Strike Rate of Batters

```python
unique_batters = np.unique(batter)

batter_total_runs = [
    int(batsman_runs[batter == player].sum())
    for player in unique_batters
]

batter_total_balls = [
    int(np.sum(batter == player))
    for player in unique_batters
]

batter_strike_rate = [
    (str(player), round((runs / balls) * 100, 2))
    for player, runs, balls in zip(unique_batters, batter_total_runs, batter_total_balls)
]

print(batter_strike_rate[:5])
```

Task 4 : Bowler Economy

```python
# Total runs scored on each delivery
total_runs = data[:, 11].astype(int)

# Unique bowlers
unique_bowlers = np.unique(bowler)

# Runs conceded by each bowler
runs_conceded = [
    int(total_runs[bowler == bowler_name].sum())
    for bowler_name in unique_bowlers
]

# Balls bowled by each bowler
balls_bowled = [
    int((bowler == bowler_name).sum())
    for bowler_name in unique_bowlers
]

# Economy rate = runs conceded / overs bowled
# Overs bowled = balls bowled / 6
economy_rate = [
    (str(bowler_name), round((runs / balls) * 6, 2))
    for bowler_name, runs, balls in zip(unique_bowlers, runs_conceded, balls_bowled)
]

print(economy_rate[200:250])
```
