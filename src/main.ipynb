Task 0 : Loading the dataset

```python
import numpy as np
 
data = np.genfromtxt(
    '../dataset/deliveries.csv',
    delimiter=',',
    dtype=str,
    skip_header=1
)
 
print("Shape:", data.shape)
print("First row:", data[0])
 
match_ids = data[:, 0].astype(int)
over = data[:, 1].astype(int)
batting_team = data[:, 2]
batter = data[:, 6]
bowler = data[:, 7]
batsman_runs = data[:, 9].astype(int)
```

Taks 1 : Total runs per match


```python
unique_match_ids = np.unique(match_ids)

total_runs_per_match = [(int(match_id), int(batsman_runs[match_ids == match_id].sum())) for match_id in unique_match_ids]

print(total_runs_per_match[:5])
```

Task 2 : Top 5 Batters


```python
# Sum runs for each batter
top_5_batters = [
    (str(player), int(batsman_runs[batter == player].sum()))
    for player in np.unique(batter)
]

# Sort by total runs in descending order and keep the top 5
top_5_batters = sorted(top_5_batters, key=lambda x: x[1], reverse=True)[:5]

print(top_5_batters)
```

Task 3 : Strike Rate of Batters

```python
unique_batters = np.unique(batter)

batter_total_runs = [
    int(batsman_runs[batter == player].sum())
    for player in unique_batters
]

batter_total_balls = [
    int(np.sum(batter == player))
    for player in unique_batters
]

batter_strike_rate = [
    (str(player), round((runs / balls) * 100, 2))
    for player, runs, balls in zip(unique_batters, batter_total_runs, batter_total_balls)
]

print(batter_strike_rate[:5])
```

Task 4 : Bowler Economy

```python
# Total runs scored on each delivery
total_runs = data[:, 11].astype(int)

# Unique bowlers
unique_bowlers = np.unique(bowler)

# Runs conceded by each bowler
runs_conceded = [
    int(total_runs[bowler == bowler_name].sum())
    for bowler_name in unique_bowlers
]

# Balls bowled by each bowler
balls_bowled = [
    int((bowler == bowler_name).sum())
    for bowler_name in unique_bowlers
]

# Economy rate = runs conceded / overs bowled
# Overs bowled = balls bowled / 6
economy_rate = [
    (str(bowler_name), round((runs / balls) * 6, 2))
    for bowler_name, runs, balls in zip(unique_bowlers, runs_conceded, balls_bowled)
]

print(economy_rate[200:250])
```

Task 5 : Runs Per Over

```python
over = data[:, 4].astype(int)

runs_per_over = np.array([
    round((batsman_runs[over == i].sum() / np.sum(over == i)) * 6, 2)
    if np.sum(over == i) else 0
    for i in range(20)
])

print(runs_per_over)
```

Task 6 : Boundary Analysis

```python
total_fours = int(np.sum(batsman_runs == 4))
total_sixes = int(np.sum(batsman_runs == 6))

boundary_mask = np.isin(batsman_runs, [4, 6])

team_boundary_counts = [
    (str(team_name), int(np.sum((batting_team == team_name) & boundary_mask)))
    for team_name in np.unique(batting_team)
]

team_with_most_boundaries = max(team_boundary_counts, key=lambda x: x[1])

print("Total 4s:", total_fours)
print("Total 6s:", total_sixes)
print("Team with most boundaries:", team_with_most_boundaries)
```

Taks 7 : Death Over Analysis

```python
# Death overs = 16 to 20
# In this dataset, over is 0-based, so that means values 15 to 19
death_overs_mask = (over >= 15) & (over <= 19)

total_runs_in_death_overs = int(total_runs[death_overs_mask].sum())

death_overs_by_team = [
    (str(team_name), int(total_runs[(batting_team == team_name) & death_overs_mask].sum()))
    for team_name in np.unique(batting_team)
]

team_with_highest_death_overs_runs = max(death_overs_by_team, key=lambda x: x[1])

print("Total runs in death overs:", total_runs_in_death_overs)
print("Team with highest death-over runs:", team_with_highest_death_overs_runs)
```

Taks 8 : High Scoring matches

```python
match_total_runs = [
    (int(match_id), int(total_runs[match_ids == match_id].sum()))
    for match_id in np.unique(match_ids)
]

highest_scoring_match = max(match_total_runs, key=lambda x: x[1])

print(highest_scoring_match)
```

Task 9: Runs per Team per Match

```python
runs_per_team_per_match = [
    (
        int(match_id),
        str(team_name),
        int(total_runs[(match_ids == match_id) & (batting_team == team_name)].sum())
    )
    for match_id in np.unique(match_ids)
    for team_name in np.unique(batting_team[match_ids == match_id])
]

print(runs_per_team_per_match[:10])
```

Task 10: Match Winner Approximation

```python
match_winner_approximation = [
    (
        int(match_id),
        str(teams[0]) if len(teams) == 1 else (
            str(teams[0]) if runs_0 > runs_1 else str(teams[1]) if runs_1 > runs_0 else "Tie"
        )
    )
    for match_id in np.unique(match_ids)
    for teams in [np.unique(batting_team[match_ids == match_id])]
    for runs_0, runs_1 in [(
        int(total_runs[(match_ids == match_id) & (batting_team == teams[0])].sum()),
        int(total_runs[(match_ids == match_id) & (batting_team == teams[1])].sum()) if len(teams) > 1 else 0
    )]
]

print(match_winner_approximation[:10])
```

Task 11: Match Scorecard Generation

```python
for match_id in np.unique(match_ids):
    teams = np.unique(batting_team[match_ids == match_id])

    print(f"Match {match_id}:")
    for team_name in teams[:10]:
        team_runs = int(total_runs[(match_ids == match_id) & (batting_team == team_name)].sum())
        print(f" {team_name}: {team_runs} runs")
    print()
```
